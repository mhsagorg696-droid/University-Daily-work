```python
import cv2

image_path = "sample.jpg"
img = cv2.imread(image_path)

print(type(img))
```

```python
import numpy as np

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

if gray.mean() > 127:
    print("Bright Image")
else:
    print("Dark Image")
```

```python
height, width = gray.shape

for i in range(0, height, 50):
    for j in range(0, width, 50):
        print(gray[i][j], end=" ")
    print()
```
