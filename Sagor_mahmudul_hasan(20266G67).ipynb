### Scikit-learn: A Concise Overview

**Definition:**
Scikit-learn is a versatile open-source Python library for machine learning. Built on NumPy and SciPy, it offers robust implementations of both supervised and unsupervised learning algorithms.

**Key Features:**
*   **Simple API:** Consistent `fit()`, `predict()`, `transform()` interface.
*   **Algorithm Variety:** Includes SVM, KNN, Decision Trees, K-Means, and more.
*   **Utility Tools:** Provides data preprocessing, model evaluation, and hyperparameter tuning capabilities.

### Scikit-learn Example: Iris Species Classification

We'll perform a basic classification task using the loaded `iris_df`. The steps will be:
1.  Prepare the data (features and target).
2.  Split the data into training and testing sets.
3.  Train a Logistic Regression model.
4.  Make predictions and evaluate the model's accuracy.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Prepare the data: Features (X) and Target (y)
X = iris_df.drop('species', axis=1) # Features are all columns except 'species'
y = iris_df['species']             # Target is the 'species' column

# 2. Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42 # 70% for training, 30% for testing
)

print(f"Training set size: {len(X_train)} samples")
print(f"Testing set size: {len(X_test)} samples")
```

```python
# 3. Initialize and train a Logistic Regression model
model = LogisticRegression(max_iter=200) # Increased max_iter for convergence
model.fit(X_train, y_train)

print("Model training complete.")
```

```python
# 4. Make predictions on the test set
y_pred = model.predict(X_test)

# 5. Evaluate the model's accuracy
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy:.2f}")
```

## Definition of Classification

In machine learning, **classification** is a supervised learning technique used to predict a **discrete category or class** for a given input. It learns from labeled data to assign new, unseen data points into one of several predefined classes.

### Applications of Classification

*   **Spam Detection:** Identifying emails as 'spam' or 'not spam'.
*   **Medical Diagnosis:** Classifying diseases (e.g., 'diabetic' or 'non-diabetic').
*   **Image Recognition:** Categorizing objects in images (e.g., 'cat', 'dog').
*   **Sentiment Analysis:** Determining sentiment of text (e.g., 'positive', 'negative').
*   **Fraud Detection:** Flagging transactions as 'fraudulent' or 'legitimate'.

### Common Classification Algorithms

*   Logistic Regression
*   K-Nearest Neighbors (KNN)
*   Support Vector Machines (SVM)
*   Decision Trees
*   Random Forest
*   Naive Bayes

```python
from sklearn.ensemble import RandomForestClassifier

# Initialize a Random Forest Classifier model
rf_model = RandomForestClassifier(random_state=42) # Using random_state for reproducibility

# Train the model
rf_model.fit(X_train, y_train)

print("Random Forest Model training complete.")
```

```python
# Make predictions on the test set
y_pred_rf = rf_model.predict(X_test)

# Evaluate the model's accuracy
accuracy_rf = accuracy_score(y_test, y_pred_rf)

print(f"Random Forest Model Accuracy: {accuracy_rf:.2f}")
```

### Regression Example: Simple Linear Regression

i just create a synthetic dataset and use `LinearRegression` to predict a continuous target variable.

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import pandas as pd

# 1. Generate a synthetic dataset for regression
X_reg, y_reg = make_regression(n_samples=100, n_features=1, noise=20, random_state=42)

# Convert to DataFrame for easier handling (optional)
X_reg_df = pd.DataFrame(X_reg, columns=['Feature'])
y_reg_df = pd.Series(y_reg, name='Target')

print(f"Generated {len(X_reg_df)} samples with 1 feature.")
```

```python
# 2. Split the data into training and testing sets
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg_df, y_reg_df, test_size=0.3, random_state=42
)

print(f"Regression Training set size: {len(X_train_reg)} samples")
print(f"Regression Testing set size: {len(X_test_reg)} samples")
```

```python
# 3. Initialize and train a Linear Regression model
reg_model = LinearRegression()
reg_model.fit(X_train_reg, y_train_reg)

print("Linear Regression model training complete.")
print(f"Coefficient: {reg_model.coef_[0]:.2f}")
print(f"Intercept: {reg_model.intercept_:.2f}")
```

```python
# 4. Make predictions on the test set
y_pred_reg = reg_model.predict(X_test_reg)

# 5. Evaluate the model
mse = mean_squared_error(y_test_reg, y_pred_reg)
r2 = r2_score(y_test_reg, y_pred_reg)

print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"R-squared (R2) Score: {r2:.2f}")
```

```python
# 6. Visualize the results
plt.figure(figsize=(10, 6))
plt.scatter(X_test_reg, y_test_reg, color='blue', label='Actual values')
plt.plot(X_test_reg, y_pred_reg, color='red', linewidth=2, label='Predicted line')
plt.xlabel('Feature')
plt.ylabel('Target')
plt.title('Linear Regression: Actual vs. Predicted Values')
plt.legend()
plt.grid(True)
plt.show()
```

## Definition of Clustering

In machine learning, **clustering** is an **unsupervised learning** technique used to group a set of data points into subsets (clusters) such that data points within the same cluster are more similar to each other than to those in other clusters. Unlike supervised learning, clustering does not rely on labeled output data; it discovers inherent structures or patterns in the data.

### Applications of Clustering

*   Customer Segmentation
*   Document Analysis
*   Image Segmentation
*   Anomaly Detection
*   Genomic Analysis
*   Recommendation Systems.

### Common Clustering Algorithms

*   K-Means
*   Hierarchical Clustering
*   DBSCAN (Density-Based Spatial Clustering of Applications with Noise)
*   Mean-Shift
*   Gaussian Mixture Models (GMM)

### Clustering Example: K-Means Clustering

We'll generate a synthetic dataset and apply K-Means clustering to group the data points.

```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import pandas as pd

# 1. Generate a synthetic dataset for clustering
X_clus, y_clus = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=42)

# Convert to DataFrame for easier handling (optional)
X_clus_df = pd.DataFrame(X_clus, columns=['Feature_1', 'Feature_2'])

print(f"Generated {len(X_clus_df)} samples for clustering.")
```

```python
# 2. Initialize and apply K-Means clustering
# We'll assume 4 clusters, as we generated the data with 4 centers
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10) # n_init is set to suppress a warning
kmeans.fit(X_clus_df)

# Get cluster labels and cluster centers
cluster_labels = kmeans.labels_
cluster_centers = kmeans.cluster_centers_

print("K-Means clustering complete.")
print(f"First 5 cluster labels: {cluster_labels[:5]}")
```

```python
# 3. Visualize the clustering results
plt.figure(figsize=(10, 7))

# Plot data points, colored by cluster
plt.scatter(X_clus_df['Feature_1'], X_clus_df['Feature_2'], c=cluster_labels, cmap='viridis', s=50, alpha=0.8)

# Plot cluster centers
plt.scatter(cluster_centers[:, 0], cluster_centers[:, 1], s=200, c='red', marker='X', edgecolor='black', label='Cluster Centers')

plt.title('K-Means Clustering Results')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```

###  Simple Classification Example: K-Nearest Neighbors (KNN) using iris dataset.

Let's apply the K-Nearest Neighbors algorithm to the Iris dataset. KNN is a non-parametric, lazy learning algorithm that classifies a new data point based on the majority class among its 'k' nearest neighbors in the feature space.

```python
from sklearn.neighbors import KNeighborsClassifier

# 1. Initialize and train a K-Nearest Neighbors model
# We'll start with k=5 (looking at 5 nearest neighbors)
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)

print("K-Nearest Neighbors model training complete.")
```

```python
# 2. Make predictions on the test set
y_pred_knn = knn_model.predict(X_test)

# 3. Evaluate the model's accuracy
accuracy_knn = accuracy_score(y_test, y_pred_knn)

print(f"K-Nearest Neighbors Model Accuracy: {accuracy_knn:.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a figure with two subplots side-by-side
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Actual Species of Test Data
sns.scatterplot(
    x=X_test['petal length (cm)'],
    y=X_test['petal width (cm)'],
    hue=y_test,
    palette='viridis',
    marker='o',
    s=100,
    ax=axes[0],
    legend='full'
)
axes[0].set_title('Actual Species in Test Set')
axes[0].set_xlabel('Petal Length (cm)')
axes[0].set_ylabel('Petal Width (cm)')
axes[0].grid(True)

# Plot 2: Predicted Species by KNN Model
sns.scatterplot(
    x=X_test['petal length (cm)'],
    y=X_test['petal width (cm)'],
    hue=y_pred_knn,
    palette='viridis',
    marker='X',
    s=100,
    ax=axes[1],
    legend='full'
)
axes[1].set_title('KNN Predicted Species on Test Set')
axes[1].set_xlabel('Petal Length (cm)')
axes[1].set_ylabel('Petal Width (cm)')
axes[1].grid(True)

plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()
```
